```python
# Setup: get the latest repo files and load the data. Run this first, every session.
import os, subprocess
REPO_URL = "https://github.com/haydenhollar/ai301-vol-gig.git"
if os.path.exists("/content/repo"):
    subprocess.run(["git", "-C", "/content/repo", "pull"], check=True)
else:
    subprocess.run(["git", "clone", REPO_URL, "/content/repo"], check=True)

import pandas as pd
DATA = "/content/repo/data"
students = pd.read_csv(f"{DATA}/students.csv", dtype=str).fillna("")
print("students:", students.shape)
for name in ["postings", "test_set"]:
    path = f"{DATA}/{name}.csv"
    if os.path.exists(path):
        globals()[name] = pd.read_csv(path, dtype=str).fillna("")
        print(name + ":", globals()[name].shape)
    else:
        print(name + ": not in the repo yet")
```

```python
assert len(students) == students.student_id.nunique(), "duplicate student IDs"
counts = students.major.value_counts()
print("students:", len(students), "| majors:", len(counts), "| fewest per major:", counts.min())
assert counts.min() >= 2, "a major has only one student"
```

```python
def slots(s): return [x.strip() for x in s.split(";") if x.strip()]
def free(row, slot): return slot in slots(row.availability)
def has(row, tag): return tag in row.skill_tags

print("J01 lawn mowing tag + Sat morning:", [r.student_id for r in students.itertuples() if has(r, "lawn mowing") and free(r, "Sat morning")])
print("J15 free Sun early morning:", [r.student_id for r in students.itertuples() if free(r, "Sun early morning")])
print("J18 any electrician license:", [r.student_id for r in students.itertuples() if "electric" in r.licenses.lower()])
print()
print(pd.read_csv(f"{DATA}/planted_cases.csv")[["posting_id", "should_match", "close_but_wrong"]].to_string(index=False))
```
